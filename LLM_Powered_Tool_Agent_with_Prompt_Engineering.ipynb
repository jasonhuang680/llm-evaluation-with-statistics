```python
from langchain_ollama import OllamaLLM
import pandas as pd
from scipy import stats

# -------------------------------
# Select your local Ollama model
# -------------------------------
llm = OllamaLLM(model="gemma3:4b", temperature=0.3)
# To use llama3.1:8b, uncomment the next line:
# llm = OllamaLLM(model="llama3.1:8b", temperature=0.3)

# -------------------------------
# Tool functions
# -------------------------------
def read_sales_data():
    """Simulate reading last quarter sales data for USA"""
    data = {
        'region': ['USA']*8,
        'month': ['2025-10','2025-11','2025-12','2025-01','2025-02','2025-03','2025-04','2025-05'],
        'sales': [120, 115, 108, 95, 88, 82, 90, 85]
    }
    df = pd.DataFrame(data)
    return df

def perform_ttest(group1, group2):
    """Perform t-test between two groups"""
    t_stat, p_value = stats.ttest_ind(group1, group2)
    return t_stat, p_value

# -------------------------------
# Simulated ReAct Agent workflow
# -------------------------------
def agent_run(question):
    # 1️⃣ Read sales data
    df = read_sales_data()
    
    # 2️⃣ Calculate key metrics
    avg_sales = df['sales'].mean()
    min_sales = df['sales'].min()
    
    # 3️⃣ Perform t-test
    group1 = df['sales'][:5].tolist()
    group2 = [135, 130, 125, 118, 110]  # Comparison group
    t_stat, p_value = perform_ttest(group1, group2)
    
    # 4️⃣ Construct prompt for the LLM
    prompt = f"""
I have a set of sales data:
{df.to_string(index=False)}

Average sales: {avg_sales:.2f}, Minimum sales: {min_sales}
t-test results: t={t_stat:.2f}, p={p_value:.4f}

Based on this information, please provide 3 actionable recommendations to improve sales, in English.
"""
    
    # 5️⃣ Call the Ollama LLM
    output = llm.generate([prompt])
    response = output.generations[0][0].text
    return response

# -------------------------------
# Run the Agent
# -------------------------------
question = "Please analyze the reason for the sales decline in USA last quarter and provide 3 actionable recommendations."
result = agent_run(question)
print("\n=== Analysis Result ===\n")
print(result)
```
